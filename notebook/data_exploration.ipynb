```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.read_csv("../archive/Data/features_30_sec.csv")

X = df[["tempo"]]
y = df["spectral_centroid_mean"]

model = LinearRegression()
model.fit(X, y)

print("Slope (m):", model.coef_)
print("Intercept (c):", model.intercept_)

new_data = pd.DataFrame({"tempo": [15]})
print("Prediction:", model.predict(new_data))
```

```python
import matplotlib.pyplot as plt

y_pred = model.predict(X)

plt.figure(figsize=(10,6))
plt.scatter(X, y, color="blue", alpha=0.5, label="Actual Data")
plt.plot(X, y_pred, color="red", linewidth=2, label="Regression Line")

plt.title("Linear Regression: Tempo vs Spectral Centroid")
plt.xlabel("Tempo")
plt.ylabel("Spectral Centroid Mean")
plt.legend()

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../archive/Data/features_30_sec.csv")

# Tempo = speed of the music (beats per minute)
# Spectral Centroid = brightness of the music

X = df["tempo"].values
Y = df["spectral_centroid_mean"].values

X = (X - X.mean()) / X.std()

m = 0
b = 0

learning_rate = 0.01
epochs = 1000
n = len(X)

loss_history = []

for epoch in range(epochs):

    Y_pred = m * X + b

    loss = np.mean((Y - Y_pred) ** 2)
    loss_history.append(loss)

    dm = (-2/n) * np.sum(X * (Y - Y_pred))
    db = (-2/n) * np.sum(Y - Y_pred)

    m = m - learning_rate * dm
    b = b - learning_rate * db

print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
plt.figure(figsize=(8,5))

plt.scatter(X, Y, alpha=0.5)

plt.plot(X, m*X+b, color='red', linewidth=3)

plt.title("Linear Regression using Gradient Descent")
plt.xlabel("Tempo (Normalized)")
plt.ylabel("Spectral Centroid Mean")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(loss_history)

plt.title("Loss during Gradient Descent")
plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")

plt.show()
```
