```python
#librosa is library used to read audio files for python

# with librosa we convert song into number => signal, sr = librosa.load(file_path, sr=22050)
# sr means sample rate => 22050 values every second
# This list is called
# Signal (Waveform)
# Every number represents the loudness of the sound at one instant.

# A 30-second song contains
# intro
# chorus
# guitar solo
# ending
# Different parts sound different.
# Instead of learning one huge song,
# we split it.
# 30 seconds => 3 sec+ 3 sec + .....
# This is called
# Segmentation

# For music,
# we describe
# Brightness
# Rhythm
# Melody
# Timbre
# These descriptions are called
# Audio Features

# 1. MFCC (Mel-Frequency Cepstral Coefficients) ✅
# "What does this sound like?"
# It captures the texture (timbre) of the sound.
# For example:
# 🎸 Guitar
# 🎹 Piano
# 🎺 Trumpet
# 🎤 Human voice
# Even if they play the same note, they sound different.
# MFCC captures that difference.
# 13 output for each segment

# 2. Chroma Features ✅
# "Which musical notes are being played?"
# Instead of instrument, it focuses on music notes.
# Western music has only 12 note classes:
# C
# C#
# D
# D#
# E
# F
# F#
# G
# G#
# A
# A#
# B
# Chroma tells us how much of each note exists.

# 3. Spectral Centroid ✅
# "Is this sound bright or dark?"
# It measures where most of the sound energy lies.
# Low frequencies
# Bass
# Drums
# Deep voice => Dark sound
# High frequencies
# Cymbals
# Flute
# Hi-hats => Bright sound

# Song (.wav)
#         │
# Load using librosa
#         │
# Split into 10 segments
#         │
# For each segment
#         │
#         ├── MFCC
#         │      → Sound texture (instrument/timbre)
#         │
#         ├── Chroma
#         │      → Musical notes & chords
#         │
#         └── Spectral Centroid
#                → Brightness of sound

# | Feature               | Easy way to remember                                         |
# | --------------------- | ------------------------------------------------------------ |
# | **MFCC**              | Instrument or sound texture                                  |
# | **Chroma**            | Musical notes/chords                                         |
# | **Spectral Centroid** | Average brightness (center of frequencies)                   |
# | **Spectral Rolloff**  | Frequency below which **85% of the sound energy** is present |
# | **zero crossing rate**| smoothness of song less zcr more smooth                      |

```

```python
import os
import json
import librosa
import numpy as np

# Dataset path
DATASET_PATH = "archive/Data/genres_original"

# Output JSON file
JSON_PATH = "data.json"

# MFCC constants
NUM_MFCC = 13
N_FFT = 2048
HOP_LENGTH = 512

# Audio processing constants
SAMPLE_RATE = 22050
TRACK_DURATION_SECONDS = 30
NUM_SEGMENTS = 10

# Samples in one track
SAMPLES_PER_TRACK = SAMPLE_RATE * TRACK_DURATION_SECONDS
# Samples in each segment
NUM_SAMPLES_PER_SEGMENT = SAMPLES_PER_TRACK // NUM_SEGMENTS

def process_dataset(dataset_path, json_path):

    data = {
        "mapping": [],
        "labels": [],
        "features": []
    }

    print("Starting feature extraction...\n")

    for i, (dirpath, dirnames, filenames) in enumerate(os.walk(dataset_path)):

        if dirpath == dataset_path:
            continue

        genre = os.path.basename(dirpath)

        print(f"\nProcessing genre: {genre}")

        data["mapping"].append(genre)

        # Loop through every audio file
        for filename in filenames:
            if filename.endswith(".wav"):
        
                file_path = os.path.join(dirpath, filename)
        
                try:
                    # Load audio
                    signal, sr = librosa.load(file_path, sr=SAMPLE_RATE)
                
                    # Check signal length
                    if len(signal) >= SAMPLES_PER_TRACK:
                
                        # Divide the signal into segments
                        for s in range(NUM_SEGMENTS):

                            start_sample = s * NUM_SAMPLES_PER_SEGMENT
                            end_sample = start_sample + NUM_SAMPLES_PER_SEGMENT
                        
                            segment = signal[start_sample:end_sample]
                            
                            # Extract MFCCs
                            mfccs = librosa.feature.mfcc(
                                y=segment,
                                sr=sr,
                                n_mfcc=NUM_MFCC,
                                n_fft=N_FFT,
                                hop_length=HOP_LENGTH
                            )
                            
                            # Take the mean of each MFCC coefficient
                            mfccs_processed = np.mean(mfccs, axis=1)
                            
                            # Optional: print shape for verification
                            # print(mfccs_processed.shape)
                
                            # Optional: print for verification
                            # print(f"{filename} -> Segment {s+1}: {segment.shape}")

                            
                            # Extract Chroma Features
                            chroma = librosa.feature.chroma_stft(
                                y=segment,
                                sr=sr,
                                n_fft=N_FFT,
                                hop_length=HOP_LENGTH
                            )

                            # Average the chroma values across time
                            chroma_processed = np.mean(chroma, axis=1)
                            
                            # Optional: verify shape
                            # print(chroma_processed.shape)
                
                except Exception as e:
                    print(f"Error loading file {file_path}: {e}")
                    continue

    print("\nFeature extraction complete.")

if __name__ == "__main__":
    process_dataset(DATASET_PATH, JSON_PATH)
```

```python
import os
import json
import librosa
import numpy as np

# Dataset Configuration

DATASET_PATH = "archive/Data/genres_original"

# Output JSON file
JSON_PATH = "data.json"

# Feature Extraction Constants

NUM_MFCC = 13

N_FFT = 2048
HOP_LENGTH = 512

SAMPLE_RATE = 22050

TRACK_DURATION_SECONDS = 30
NUM_SEGMENTS = 10


# Samples in complete track
SAMPLES_PER_TRACK = SAMPLE_RATE * TRACK_DURATION_SECONDS

# Samples in each 3-second segment
NUM_SAMPLES_PER_SEGMENT = SAMPLES_PER_TRACK // NUM_SEGMENTS

# Feature Extraction Function

def process_dataset(dataset_path, json_path):

    data = {
        "mapping": [],
        "labels": [],
        "features": []
    }


    print("\nStarting Feature Extraction...\n")


    # Loop through genres
    for i, genre in enumerate(sorted(os.listdir(dataset_path))):

        genre_path = os.path.join(dataset_path, genre)


        # Ignore non-folder files
        if not os.path.isdir(genre_path):
            continue


        print(f"Processing Genre: {genre}")


        # Store genre name
        data["mapping"].append(genre)



        # Loop through audio files
        for filename in os.listdir(genre_path):

            if filename.endswith(".wav"):


                file_path = os.path.join(
                    genre_path,
                    filename
                )


                try:

                    print(f"  Processing: {filename}")


                    # Load audio
                    signal, sr = librosa.load(
                        file_path,
                        sr=SAMPLE_RATE
                    )


                    # Check duration
                    if len(signal) < SAMPLES_PER_TRACK:
                        print("  Skipped (short file)")
                        continue

                    # Segment Extraction

                    for s in range(NUM_SEGMENTS):


                        start_sample = (
                            s * NUM_SAMPLES_PER_SEGMENT
                        )


                        end_sample = (
                            start_sample +
                            NUM_SAMPLES_PER_SEGMENT
                        )


                        segment = signal[
                            start_sample:end_sample
                        ]

                        # MFCC

                        mfccs = librosa.feature.mfcc(
                            y=segment,
                            sr=sr,
                            n_mfcc=NUM_MFCC,
                            n_fft=N_FFT,
                            hop_length=HOP_LENGTH
                        )


                        mfccs_processed = np.mean(
                            mfccs,
                            axis=1
                        )

                        # Chroma

                        chroma = librosa.feature.chroma_stft(
                            y=segment,
                            sr=sr,
                            n_fft=N_FFT,
                            hop_length=HOP_LENGTH
                        )


                        chroma_processed = np.mean(
                            chroma,
                            axis=1
                        )

                        # Spectral Centroid

                        spectral_centroid = librosa.feature.spectral_centroid(
                            y=segment,
                            sr=sr,
                            n_fft=N_FFT,
                            hop_length=HOP_LENGTH
                        )


                        spectral_centroid_processed = np.mean(
                            spectral_centroid
                        )

                        # Spectral Rolloff

                        spectral_rolloff = librosa.feature.spectral_rolloff(
                            y=segment,
                            sr=sr,
                            n_fft=N_FFT,
                            hop_length=HOP_LENGTH
                        )


                        spectral_rolloff_processed = np.mean(
                            spectral_rolloff
                        )

                        # Zero Crossing Rate

                        zcr = librosa.feature.zero_crossing_rate(
                            y=segment,
                            hop_length=HOP_LENGTH
                        )


                        zcr_processed = np.mean(zcr)

                        # Combine Features

                        feature_vector = np.hstack(
                            (
                                mfccs_processed,
                                chroma_processed,
                                spectral_centroid_processed,
                                spectral_rolloff_processed,
                                zcr_processed
                            )
                        )



                        # Save feature vector
                        data["features"].append(
                            feature_vector.tolist()
                        )


                        # Save genre label
                        data["labels"].append(i)



                except Exception as e:

                    print(
                        f"Error processing {file_path}: {e}"
                    )

                    continue

    # Save JSON

    with open(json_path, "w") as f:

        json.dump(
            data,
            f,
            indent=4
        )


    print("Feature Extraction Completed")


    print(
        "Total Feature Vectors:",
        len(data["features"])
    )


    print(
        "Total Labels:",
        len(data["labels"])
    )


    print(
        "Feature Vector Size:",
        len(data["features"][0])
    )


# Main Execution

if __name__ == "__main__":

    process_dataset(
        DATASET_PATH,
        JSON_PATH
    )
```
